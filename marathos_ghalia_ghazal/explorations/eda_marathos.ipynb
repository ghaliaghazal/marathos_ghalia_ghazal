```python
# Läs in rawdata från Volume
df = spark.read.csv("/Volumes/marathos/default/raw/TWO_CENTURIES_OF_UM_RACES 2.csv", header=True, inferSchema=True)

display(df)
```

Antal rader

```python
df.count()
```

```python
len(df.columns)
```

Kontrollera schemat

```python
df.printSchema()
```

Beskrivande sammanfattning av numeriska fält

```python
display(df.describe())
```

Hitta Null-värden (tomma värden)

```python
from pyspark.sql.functions import col, sum 
null_counts = df.select([sum(col(c).isNull().cast("int")).alias(c) for c in df.columns]) 

display(null_counts) 
```

Antal unika lopp (Event)

```python
df.select("Event name").distinct().count()
```

Åldersfördelningen bland löparna

```python
df.select("Athlete year of birth").distinct().count()
```

```python
df.groupBy("Athlete year of birth").count().orderBy("Athlete year of birth")

display(df)
```

Mest representerade länder 

```python
df.groupBy("Athlete country").count().orderBy("count", ascending=False)
display(df)
```

```python
display(df.groupBy("Athlete country").count().orderBy("count", ascending=False).limit(10))
```
