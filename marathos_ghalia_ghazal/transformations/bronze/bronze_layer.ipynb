```python
# Läs in rawdata från Volume

df = spark.read.csv("/Volumes/marathos/default/raw/TWO_CENTURIES_OF_UM_RACES 2.csv", header=True, inferSchema=True)

display(df)
```

```python
# Kontrollera hur det ser ut
df.printSchema()
```

```python
# Byt ut mellanslag och specialtecken i kolumnnamnen

new_columns = [c.replace(" ", "_").replace("/", "_") for c in df.columns]
df = df.toDF(*new_columns)
```

```python
# Spara till bronze lagret

df.write.format("delta").mode("overwrite").saveAsTable("marathos.bronze.raw_data")

print("✅ Bronze-tabell skapad!")
```
